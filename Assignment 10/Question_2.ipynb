## Q1. Structural Probability and Properties

The Beta distribution is commonly used in Bayesian statistics to represent uncertainty about probabilities that lie between 0 and 1. Its probability density function is given by:

> ($f(\theta)=\dfrac{1}{B(\alpha,\beta)}\theta^{\alpha-1}(1-\theta)^{\beta-1}, ; 0 \le \theta \le 1$)

where:

* ($\alpha$) represents the number of prior successes (clicks).
* ($\beta$) represents the number of prior failures (non-clicks).
* ($B(\alpha,\beta)$) is the Beta function used for normalization.

The three parameter settings illustrate different prior beliefs:

1.  **Beta(1,1) – Uninformative Prior**

    *   This distribution is uniform over the interval ([0,1]).
    *   All click-through rates are considered equally likely before observing any data.

2.  **Beta(2,8) – Right-Skewed Distribution**

    *   The density is concentrated toward smaller values of ($\theta$).
    *   This reflects a prior belief that the advertisement is unlikely to receive many clicks.
    *   The center of mass shifts toward 0.

3.  **Beta(8,2) – Left-Skewed Distribution**

    *   The density is concentrated toward larger values of ($\theta$).
    *   This reflects a prior belief that the advertisement has a relatively high click-through rate.
    *   The center of mass shifts toward 1.

The balance between ($\alpha$) and ($\beta$) determines where the density is concentrated. Larger values of ($\alpha$) favor higher click probabilities, while larger values of ($\beta$) favor lower click probabilities.

## Q2. Sequential Likelihood and Joint History

Each user interaction is modeled as an independent Bernoulli trial.

The likelihood contribution of a single observation ($y_k$) is:

> ($L(y_k|\theta)=\theta^{y_k}(1-\theta)^{1-y_k}$)

where:

* ($y_k=1$) if the advertisement is clicked.
* ($y_k=0$) if the advertisement is not clicked.

The likelihood behaves as follows:

* If ($y_k=1$), then ($L(y_k|\theta)=\theta$).
* If ($y_k=0$), then ($L(y_k|\theta)=1-\theta$).

Assuming that all user interactions are conditionally independent given the click-through rate, the joint likelihood function for the running history vector is:

> ($L(\mathbf y^{(k)}|\theta)=\prod_{i=1}^{k}\theta^{y_i}(1-\theta)^{1-y_i}$)

This can also be expressed as:

> ($L(\mathbf y^{(k)}|\theta)=\theta^{\sum y_i}(1-\theta)^{k-\sum y_i}$)

where:

* ($\sum y_i$) is the total number of clicks.
* ($k-\sum y_i$) is the total number of non-clicks.

## Q3. Closed-Form Analytical Updates (Conjugacy)

Initially, the prior distribution is assumed to be:

> ($\Theta \sim Beta(\alpha_0,\beta_0)$)

with density:

> ($f(\theta)=\dfrac{1}{B(\alpha_0,\beta_0)}\theta^{\alpha_0-1}(1-\theta)^{\beta_0-1}$)

By Bayes' theorem, the posterior distribution at time step ($k$) is proportional to:

> ($f(\theta|\mathbf y^{(k)}) \propto L(y_k|\theta)\times f(\theta|\mathbf y^{(k-1)})$)

Suppose that at step ($k-1$):

> ($\Theta|\mathbf y^{(k-1)} \sim Beta(\alpha_{k-1},\beta_{k-1})$)

Multiplying the likelihood and the prior gives:

> ($f(\theta|\mathbf y^{(k)}) \propto \theta^{y_k}(1-\theta)^{1-y_k}\theta^{\alpha_{k-1}-1}(1-\theta)^{\beta_{k-1}-1}$)

Combining the exponents yields:

> ($f(\theta|\mathbf y^{(k)}) \propto \theta^{\alpha_{k-1}+y_k-1}(1-\theta)^{\beta_{k-1}+(1-y_k)-1}$)

Therefore, the posterior distribution remains a Beta distribution:

> ($\Theta|\mathbf y^{(k)} \sim Beta(\alpha_k,\beta_k)$)

with the recursive updates:

> ($\boxed{\alpha_k=\alpha_{k-1}+y_k}$)

> ($\boxed{\beta_k=\beta_{k-1}+1-y_k}$)

Thus:

* Every click increments ($\alpha$) by one.
* Every non-click increments ($\beta$) by one.

This property is known as **Beta-Binomial conjugacy**, which allows posterior updates to be performed analytically without numerical integration.

### Posterior Mean

The posterior mean at step ($k$) is:

> ($\boxed{\mathbb E[\Theta|\mathbf y^{(k)}]=\dfrac{\alpha_k}{\alpha_k+\beta_k}}$)

The posterior mean provides the Bayesian estimate of the advertisement's click-through rate after observing the available evidence.

## Q4. Dynamic Shifting Mechanics

The Beta distribution is updated analytically after every observation.

### Case 1: Click Observed (($y_k=1$))

If a click is observed:

> ($\alpha_k=\alpha_{k-1}+1$)

while ($\beta_k$) remains unchanged.

As a result:

* The posterior distribution shifts toward larger values of ($\theta$).
* The posterior mean increases.
* The peak of the density moves toward higher click-through rates.

### Case 2: No Click Observed (($y_k=0$))

If no click is observed:

> ($\beta_k=\beta_{k-1}+1$)

while ($\alpha_k$) remains unchanged.

Consequently:

* The posterior distribution shifts toward smaller values of ($\theta$).
* The posterior mean decreases.
* The peak of the density moves toward lower click-through rates.

### Comparison with Non-Conjugate Models

In conjugate Bayesian models such as the Beta-Binomial model:

* Posterior distributions can be updated analytically.
* No numerical approximation methods are required.
* Computation is extremely efficient.

In contrast, non-conjugate models such as the 2PL Item Response Theory model do not admit closed-form posterior distributions. In those cases:

* Numerical grid approximation methods are required.
* Posterior normalization must be performed numerically.
* Bayesian updates are computationally more expensive.

Therefore, the Beta-Binomial model is particularly attractive for real-time applications that require rapid sequential updates.

## Q5. Running Point Estimators

The two commonly used Bayesian point estimators are the Posterior Mean and the Maximum A Posteriori (MAP) estimate.

### Running Posterior Mean

The posterior mean is given by:

> ($\boxed{\widehat{\theta}_{Bayes}^{(k)}=\dfrac{\alpha_k}{\alpha_k+\beta_k}}$)

This estimate minimizes the mean squared error under Bayesian decision theory.

---

### Running MAP Estimate

For a Beta distribution with:

> ($\alpha_k>1$) and ($\beta_k>1$)

the MAP estimate is:

> ($\boxed{\widehat{\theta}_{MAP}^{(k)}=\dfrac{\alpha_k-1}{\alpha_k+\beta_k-2}}$)

Special cases occur when either parameter is less than or equal to one. In such cases, the density is maximized at one of the boundaries:

* MAP = 0 if ($\alpha_k \le 1$)
* MAP = 1 if ($\beta_k \le 1$)

For most practical situations involving sufficient observations, the closed-form expression above can be used directly.

## Q6. Performance Tracking and Convergence Analysis

In this simulation, the true click-through rate of the advertisement is assumed to be:

> ($\theta_{true}=0.35$)

The prior distribution is initialized as:

> ($Beta(1,1)$)

which represents complete initial uncertainty regarding the advertisement's performance.

At each impression:

1.  A user interaction is simulated using the true click probability of 0.35.
2.  The observed outcome is recorded as either a click or a non-click.
3.  The parameters ($\alpha_k$) and ($\beta_k$) are updated analytically.
4.  The Posterior Mean and MAP estimates are computed and stored.
5.  The estimation process is repeated until 100 impressions have been observed.

The resulting Plotly graph illustrates how both estimators evolve over time together with the true click-through rate.

### Analysis

At the beginning of the simulation, only a few observations are available. Consequently:

*   The posterior distribution is relatively wide.
*   The estimates may fluctuate significantly.
*   The influence of the prior distribution is relatively strong.

As additional user interactions are observed:

*   More evidence is accumulated.
*   The influence of the initial prior gradually diminishes.
*   Both estimators become increasingly stable.

When the number of impressions approaches 100:

*   The Posterior Mean typically converges toward 0.35.
*   The MAP estimate also converges toward 0.35.
*   The difference between the estimated values and the true click-through rate becomes smaller.

This behavior illustrates an important property of Bayesian learning: the accumulation of evidence eventually dominates the prior assumptions. Although the prior influences the early stages of estimation, sufficient data allow the model to accurately recover the underlying click-through rate.

Therefore, the convergence of both estimators toward the true parameter value demonstrates that sequential Bayesian updating provides an effective and computationally efficient method for tracking advertisement performance in real time.

# **Q1. Structural Probability and Properties (Beta PDFs)**

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import beta

# Values of theta between 0 and 1
theta = np.linspace(0.001, 0.999, 1000)

# Parameter pairs
parameters = [
    (1, 1),
    (2, 8),
    (8, 2)
]

# Create figure
fig = go.Figure()

for alpha, beta_param in parameters:
    pdf_values = beta.pdf(theta, alpha, beta_param)

    fig.add_trace(
        go.Scatter(
            x=theta,
            y=pdf_values,
            mode="lines",
            name=f"Beta({alpha}, {beta_param})"
        )
    )

fig.update_layout(
    title="Beta Distribution PDFs",
    xaxis_title="Theta (CTR)",
    yaxis_title="Probability Density",
    template="plotly_white"
)

fig.show()
```

# **Q2. Sequential Likelihood and Joint Likelihood**

**Single Observation Likelihood**

```python
# Bernoulli likelihood function

def likelihood(theta, y):
    return (theta ** y) * ((1 - theta) ** (1 - y))


# Example
theta = 0.35
y = 1

print("Single Observation Likelihood:")
print(likelihood(theta, y))
```

**Joint Likelihood**

```python
# Example click history
responses = [1, 0, 1, 1, 0]

theta = 0.35

joint_likelihood = 1

for response in responses:
    joint_likelihood *= likelihood(theta, response)

print("Joint Likelihood:")
print(joint_likelihood)
```

# **Q3. Closed-Form Beta Posterior Updates**

```python
# Initial prior parameters
alpha = 1
beta_param = 1

# Example sequence of observations
responses = [1, 0, 1, 1, 0]

print("Sequential Posterior Updates\n")

for i, y in enumerate(responses, start=1):

    # Update rules
    alpha += y
    beta_param += (1 - y)

    # Posterior mean
    posterior_mean = alpha / (alpha + beta_param)

    print(f"Step {i}")
    print(f"Alpha = {alpha}")
    print(f"Beta = {beta_param}")
    print(f"Posterior Mean = {posterior_mean:.4f}\n")
```

# **Q4. Dynamic Shifting of the Posterior Distribution**

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import beta

theta = np.linspace(0.001, 0.999, 1000)

# Prior distribution
alpha_prior = 3
beta_prior = 3

# Posterior after a click
alpha_click = alpha_prior + 1
beta_click = beta_prior

# Posterior after a non-click
alpha_no_click = alpha_prior
beta_no_click = beta_prior + 1


fig = go.Figure()

# Prior
fig.add_trace(
    go.Scatter(
        x=theta,
        y=beta.pdf(theta, alpha_prior, beta_prior),
        mode="lines",
        name="Prior"
    )
)

# Click
fig.add_trace(
    go.Scatter(
        x=theta,
        y=beta.pdf(theta, alpha_click, beta_click),
        mode="lines",
        name="After Click"
    )
)

# No Click
fig.add_trace(
    go.Scatter(
        x=theta,
        y=beta.pdf(theta, alpha_no_click, beta_no_click),
        mode="lines",
        name="After Non-Click"
    )
)

fig.update_layout(
    title="Posterior Shifting of the Beta Distribution",
    xaxis_title="Theta",
    yaxis_title="Probability Density",
    template="plotly_white"
)

fig.show()
```

# **Q5. Running Posterior Mean and MAP Estimates**

```python
# Example posterior parameters
alpha = 15
beta_param = 8

# Posterior Mean
posterior_mean = alpha / (alpha + beta_param)

# MAP estimate
if alpha > 1 and beta_param > 1:
    map_estimate = (alpha - 1) / (alpha + beta_param - 2)
else:
    map_estimate = None

print("Posterior Mean =", posterior_mean)
print("MAP Estimate =", map_estimate)
```

# **Q6. Performance Tracking and Convergence Analysis**

**Simulating 100 Advertisement Impressions**

```python
import numpy as np
import plotly.graph_objects as go

# Reproducibility
np.random.seed(42)

# True click-through rate
theta_true = 0.35

# Number of impressions
n = 100

# Initial prior
alpha = 1
beta_param = 1

# Lists for storing estimates
posterior_means = []
map_estimates = []

# Step 0 estimates
posterior_means.append(alpha / (alpha + beta_param))
map_estimates.append(0.5)


# Sequential updates
for _ in range(n):

    # Simulate user click
    u = np.random.uniform(0, 1)

    if u < theta_true:
        y = 1
    else:
        y = 0

    # Update posterior parameters
    alpha += y
    beta_param += (1 - y)

    # Posterior Mean
    posterior_mean = alpha / (alpha + beta_param)
    posterior_means.append(posterior_mean)

    # MAP estimate
    if alpha > 1 and beta_param > 1:
        map_value = (alpha - 1) / (alpha + beta_param - 2)
    elif alpha <= 1:
        map_value = 0
    else:
        map_value = 1

    map_estimates.append(map_value)
```

**Plotting the Estimator Progression**

```python
steps = np.arange(0, n + 1)

fig = go.Figure()

# Posterior Mean
fig.add_trace(
    go.Scatter(
        x=steps,
        y=posterior_means,
        mode="lines+markers",
        name="Posterior Mean"
    )
)

# MAP Estimate
fig.add_trace(
    go.Scatter(
        x=steps,
        y=map_estimates,
        mode="lines+markers",
        name="MAP Estimate"
    )
)

# True CTR
fig.add_trace(
    go.Scatter(
        x=steps,
        y=[theta_true] * (n + 1),
        mode="lines",
        name="True CTR (0.35)"
    )
)

fig.update_layout(
    title="Sequential Bayesian Estimation of Click-Through Rate",
    xaxis_title="Impression Number",
    yaxis_title="CTR Estimate",
    template="plotly_white"
)

fig.show()
```
