# Q1. Visualizing the Mechanics

The two-parameter logistic (2PL) model describes the probability that a user answers an item correctly based on their latent ability ($\theta$). The model is given by:

> $P(Y_i=1 \mid \Theta=\theta)=\dfrac{1}{1+e^{-a_i(\theta-b_i)}}$

where:

* ($a_i$) is the discrimination parameter.
* ($b_i$) is the difficulty parameter.
* ($\theta$) represents the user's latent ability.

The discrimination parameter determines how steeply the probability changes with ability. A larger value of ($a_i$) produces a steeper curve, meaning that the item is better at distinguishing between users with different ability levels. Conversely, a smaller value of ($a_i$) produces a flatter curve, making the item less informative.

The difficulty parameter controls the horizontal position of the logistic curve. Increasing the value of ($b_i$) shifts the entire curve to the right, implying that a higher ability level is required to achieve the same probability of answering correctly. Similarly, decreasing ($b_i$) shifts the curve to the left, making the item easier for users with lower ability levels.

Therefore, changing the difficulty parameter does not alter the shape of the curve; it only changes the ability level at which the user has a 50% chance of answering correctly.

# Q2. Sequential Likelihood Contribution

For a single response at step ($k$), where the user's response is either correct (($y_k=1$)) or incorrect (($y_k=0$)), the likelihood contribution is given by:

> $L(y_k \mid \theta)=p_k(\theta)^{y_k}(1-p_k(\theta))^{1-y_k}$

where ($p_k(\theta)$) is the probability of answering item ($k$) correctly under the 2PL model.

The likelihood function appropriately handles both possible outcomes:

* If the response is correct (($y_k=1$)), the likelihood reduces to ($p_k(\theta)$).
* If the response is incorrect (($y_k=0$)), the likelihood reduces to ($1-p_k(\theta)$).

Assuming that item responses are conditionally independent given the user's latent ability, the joint likelihood for all responses observed up to step ($k$) is:

> $L(\mathbf{y}^{(k)} \mid \theta)=\prod_{i=1}^{k} p_i(\theta)^{y_i}(1-p_i(\theta))^{1-y_i}$

This likelihood function summarizes all the information contained in the user's response history and is used in the Bayesian updating process.

# Q3. Mathematical Formulation of the Running Update

Bayesian estimation updates the user's ability distribution every time a new response is observed. Initially, the user's ability is assumed to follow a standard normal distribution:

> $\Theta \sim \mathcal{N}(0,1)$

After observing the first item response, the prior distribution is combined with the likelihood of that response to produce the posterior distribution.

For the general case at step ($k$), the recursive Bayesian update can be written as:

> $f_{\Theta \mid \mathbf{Y}^{(k)}}(\theta \mid \mathbf{y}^{(k)}) \propto L(y_k \mid \theta)f_{\Theta \mid \mathbf{Y}^{(k-1)}}(\theta \mid \mathbf{y}^{(k-1)})$

This means that:

1. The posterior distribution from the previous step becomes the prior distribution for the current step.
2. The likelihood contribution of the new response is calculated.
3. The prior distribution is multiplied by the likelihood.
4. The resulting values are normalized so that the posterior distribution integrates to one.

This recursive approach allows the user's ability estimate to be continuously refined as additional responses are observed.

# Q4. Dynamic Shifting of the Posterior Distribution

Suppose the user correctly answers a highly difficult item, meaning that:

* ($y_k=1$)
* ($b_k$) is relatively large.

Difficult items generally have a low probability of being answered correctly by users with lower ability levels. Therefore, a correct response provides strong evidence that the user's ability is higher than previously estimated.

Mathematically, the likelihood function assigns greater probability to larger values of ($\theta$) when a difficult question is answered correctly. When this likelihood is multiplied by the previous posterior distribution during Bayesian updating, the posterior distribution shifts toward higher ability values.

As a result:

* The peak of the posterior distribution moves to the right.
* The posterior mean increases.
* The MAP (Maximum A Posteriori) estimate increases.

In practical terms, correctly answering difficult questions causes the system to revise its estimate of the user's ability upward.

# Q5. Tracking Certainty and Sharpness

The discrimination parameter determines how informative an item is when estimating a user's ability.

When the discrimination parameter is large:

* The logistic curve becomes steeper.
* Small differences in ability produce large differences in response probability.
* The item provides more information about the user's ability.
* The posterior distribution becomes narrower and more concentrated.

A narrower posterior distribution corresponds to lower variance and greater confidence in the estimated ability level.

When the discrimination parameter is small:

* The logistic curve becomes flatter.
* Response probabilities change slowly with ability.
* The item provides less information.
* The posterior distribution remains relatively wide.

A wider posterior distribution indicates greater uncertainty regarding the user's ability estimate.

Therefore, highly discriminating questions significantly improve the precision of the Bayesian ability estimate, whereas low-discrimination questions contribute relatively little information.

# Q6. Numerical Implementation of a Running Grid

A practical numerical approach for Bayesian updating is to represent the posterior distribution using a fixed grid of ability values.

The algorithm proceeds as follows:

1. Define a fixed grid of ability values, typically ranging from (-4) to (4).

2. Initialize the prior distribution using the standard normal distribution and normalize it so that the total probability is equal to one.

3. After a user answers an item, compute the response probability for every ability value on the grid using the 2PL model.

4. Calculate the likelihood of the observed response across the entire grid.

5. Multiply the likelihood values by the current posterior distribution to obtain the unnormalized posterior distribution.

6. Compute the normalization constant numerically using an appropriate integration method (such as the trapezoidal rule).

7. Divide the unnormalized posterior by the normalization constant so that the posterior distribution sums to one.

8. Store the updated posterior distribution and use it as the prior distribution for the next item.

9. Repeat this process sequentially for all item responses.

This grid-based approach provides an efficient and intuitive way to perform Bayesian updating without requiring analytical solutions for the posterior distribution.

# **Q1. Visualizing the Mechanics (Plotly)**

```python
import numpy as np
import plotly.graph_objects as go

# Ability values
theta = np.linspace(-4, 4, 500)

# 2PL probability function
def probability(theta, a, b):
    return 1 / (1 + np.exp(-a * (theta - b)))

# Create figure
fig = go.Figure()

# Different discrimination parameters
fig.add_trace(
    go.Scatter(
        x=theta,
        y=probability(theta, 0.8, 0),
        mode="lines",
        name="a = 0.8, b = 0"
    )
)

fig.add_trace(
    go.Scatter(
        x=theta,
        y=probability(theta, 2.0, 0),
        mode="lines",
        name="a = 2.0, b = 0"
    )
)

# Same discrimination parameter with different difficulties
for b in [-1, 0, 1]:
    fig.add_trace(
        go.Scatter(
            x=theta,
            y=probability(theta, 1.5, b),
            mode="lines",
            name=f"a = 1.5, b = {b}"
        )
    )

fig.update_layout(
    title="2PL Item Response Curves",
    xaxis_title="Ability (Theta)",
    yaxis_title="P(Y=1 | Theta)",
    template="plotly_white"
)

fig.show()
```

# **Q2. Sequential Likelihood Contribution**

```python
import numpy as np

# Single-item likelihood function
def likelihood(theta, y, a, b):
    p = 1 / (1 + np.exp(-a * (theta - b)))
    return (p ** y) * ((1 - p) ** (1 - y))


# Example values
theta = 0.75
y = 1
a = 1.2
b = 0.5

L = likelihood(theta, y, a, b)

print("Likelihood contribution of the response:")
print(L)
```

```python
# Responses and item parameters
responses = [1, 0, 1]

a_values = [1.5, 0.8, 1.2]
b_values = [0.2, -0.5, 1.0]

theta = 0.75

joint_likelihood = 1

for y, a, b in zip(responses, a_values, b_values):
    joint_likelihood *= likelihood(theta, y, a, b)

print("Joint likelihood =", joint_likelihood)
```

# **Q3. Bayesian Running Posterior Update**

```python
import numpy as np
from scipy.stats import norm

# Ability grid
theta_grid = np.linspace(-4, 4, 1000)

# Standard normal prior
prior = norm.pdf(theta_grid, 0, 1)

# Normalize prior
prior /= np.trapz(prior, theta_grid)

# New response
y_k = 1
a_k = 1.5
b_k = 0.5

# Response probabilities
p = 1 / (1 + np.exp(-a_k * (theta_grid - b_k)))

# Likelihood
likelihood_values = (p ** y_k) * ((1 - p) ** (1 - y_k))

# Posterior update
posterior = prior * likelihood_values

# Normalize posterior
posterior /= np.trapz(posterior, theta_grid)

print("Posterior successfully updated.")
```

# **Q6. Numerical Implementation of the Running Grid**

```python
import numpy as np
from scipy.stats import norm

# Ability grid
theta_grid = np.linspace(-4, 4, 1000)

# Initial prior
posterior = norm.pdf(theta_grid, 0, 1)
posterior /= np.trapz(posterior, theta_grid)

# Example item
a = 1.3
b = 0.4
y = 1

# Compute probabilities
p = 1 / (1 + np.exp(-a * (theta_grid - b)))

# Compute likelihood
likelihood_values = (p ** y) * ((1 - p) ** (1 - y))

# Bayesian update
posterior *= likelihood_values

# Normalize posterior
normalization_constant = np.trapz(posterior, theta_grid)
posterior /= normalization_constant

print("Normalization constant =", normalization_constant)
print("Posterior successfully normalized.")
```

# **Q7. Running Bayesian Ability Estimation (n = 20)**

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import norm

# For reproducibility
np.random.seed(42)

# True ability
theta_true = 0.75

# Number of items
n_items = 20

# Ability grid
theta_grid = np.linspace(-4, 4, 1000)

# Initial posterior (standard normal prior)
posterior = norm.pdf(theta_grid, 0, 1)
posterior /= np.trapz(posterior, theta_grid)

# Store estimates
posterior_means = []
map_estimates = []

# Step 0 estimates
posterior_means.append(
    np.trapz(theta_grid * posterior, theta_grid)
)

map_estimates.append(
    theta_grid[np.argmax(posterior)]
)

# Sequential simulation
for k in range(n_items):

    # Random item parameters
    b_k = np.random.normal(0, 1)
    a_k = np.random.uniform(0.5, 2.0)

    # True probability of a correct response
    p_true = 1 / (1 + np.exp(-a_k * (theta_true - b_k)))

    # Simulate response
    u = np.random.uniform(0, 1)
    y_k = 1 if u < p_true else 0

    # Probability across the theta grid
    p_grid = 1 / (1 + np.exp(-a_k * (theta_grid - b_k)))

    # Likelihood values
    likelihood_values = (
        (p_grid ** y_k)
        * ((1 - p_grid) ** (1 - y_k))
    )

    # Bayesian update
    posterior *= likelihood_values

    # Normalize posterior
    posterior /= np.trapz(posterior, theta_grid)

    # Posterior mean
    posterior_mean = np.trapz(
        theta_grid * posterior,
        theta_grid
    )

    posterior_means.append(posterior_mean)

    # MAP estimate
    map_estimate = theta_grid[np.argmax(posterior)]

    map_estimates.append(map_estimate)

# Plot estimator progression
steps = np.arange(0, n_items + 1)

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=steps,
        y=posterior_means,
        mode="lines+markers",
        name="Posterior Mean"
    )
)

fig.add_trace(
    go.Scatter(
        x=steps,
        y=map_estimates,
        mode="lines+markers",
        name="MAP Estimate"
    )
)

fig.add_trace(
    go.Scatter(
        x=steps,
        y=[theta_true] * (n_items + 1),
        mode="lines",
        name="True Ability (0.75)"
    )
)

fig.update_layout(
    title="Running Bayesian Ability Estimates",
    xaxis_title="Item Number",
    yaxis_title="Ability Estimate",
    template="plotly_white"
)

fig.show()
```
